```python
import pandas as pd
import numpy as np

orders = pd.DataFrame({
    'order_id':   range(1001, 1011),
    'region':     ['Lagos','Abuja','Lagos','Kano','Abuja',
                   'Lagos','Ibadan','Kano','Abuja','Lagos'],
    'category':   ['Electronics','Home','Home','Electronics','Home',
                   'Electronics','Home','Home','Electronics','Home'],
    'units':      [10, 6, 9, 4, 13, 7, 8, 5, 12, 3],
    'unit_price': [200, 90, 65, 250, 100, 170, 60, 240, 120, 80],
})
```

```python
# Filter for rows where the region column equals 'Kano'
kano_rows = orders[orders['region'] == 'Kano']
kano_rows
```

```python
# Filter using the 'orders' dataframe
home_above_8 = orders[(orders['units'] > 8) & (orders['category'] == 'Home')]
home_above_8
```

```python
# Select rows using isin and pick specific columns using .loc
result = orders.loc[orders['region'].isin(['Lagos', 'Abuja']), ['order_id', 'units']]
result
```

```python
# Multiply units by unit_price to create the new revenue column
orders['revenue'] = orders['units'] * orders['unit_price']
orders.head()
```

```python
import numpy as np

# Apply np.where(condition, value_if_true, value_if_false)
orders['tier'] = np.where(orders['revenue'] > 1000, 'Large', 'Regular')
orders.head()
```

```python
# Use the .str.upper() accessor to convert the region names to uppercase
orders['region_upper'] = orders['region'].str.upper()
orders.head()
```

```python
# Sort descending, then reset index and drop the old scrambled index
orders = orders.sort_values(by='revenue', ascending=False)
orders = orders.reset_index(drop=True)
orders
```

```python
# Sort by revenue, take the top 3 rows, select columns, and clean the index
challenge_result = (
    orders.sort_values(by='revenue', ascending=False)
    .head(3)[['region', 'revenue']]
    .reset_index(drop=True)
)
challenge_result
```
