```python
import numpy as np
prices = np.array([120, 80, 45, 210])
units = np.array([3100, 2500, 1800, 1250])
revenue = prices * units # vectorised: no loop needed
revenue # array([372000, 200000, 81000, 262500])
revenue.sum() # 915500
revenue.mean() # 228875.0
revenue.max(), revenue.argmax() # 372000, 0 (index of the max)
revenue > 200000 # array([ True, False, False, True])
```
