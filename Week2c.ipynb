```python
# Creating a 1D array from a list
arr_1d = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr_1d)

# Creating a 2D array (matrix)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:\n", arr_2d)

# Creating an array filled with zeros
zeros_arr = np.zeros((2, 4))
print("\nZeros Array:\n", zeros_arr)

# Creating an array with a range of values
range_arr = np.arange(0, 10, 2)
print("\nRange Array:", range_arr)
```

```python
a = np.array([10, 20, 30, 40])
m = np.array([[1, 2, 3], [4, 5, 6]])

np.zeros((2, 3))        # 2x3 of zeros
np.arange(0, 10, 2)     # 0, 2, 4, 6, 8
np.linspace(0, 1, 5)    # 5 evenly spaced numbers from 0 to 1

m.shape   # (2, 3)
m.dtype   # int64
```

```python
a * 2        # [20 40 60 80]
a + 10       # [20 30 40 50]
a ** 2       # [100 400 900 1600]
a + a        # element by element
```

```python
df = pd.DataFrame({
    "region": ["Lagos", "Abuja", "Lagos", "Kano", "Abuja"],
    "product": ["A", "B", "B", "A", "A"],
    "revenue": [500, 300, 450, 200, 380],
    "cost": [300, 200, 250, 150, 210],
})
```

```python
df[df["revenue"] > 400]
df[(df["region"] == "Lagos") & (df["revenue"] > 400)]   # use & and |, with parentheses
df[df["region"].isin(["Lagos", "Kano"])]
```

```python
df["profit"] = df["revenue"].sub(df["cost"])
df["margin"] = df["profit"] / df["revenue"]
df["tier"] = np.where(df["revenue"] > 400, "High", "Low")   # NumPy inside Pandas
df = df.rename(columns={"revenue": "rev"})
df = df.drop(columns=["tier"])
```

```python
df.groupby("region")["profit"].sum()

df.groupby("region").agg(
    total_revenue=("rev", "sum"),
    avg_profit=("profit", "mean"),
    orders=("product", "count"),
).reset_index()
```

```python
managers = pd.DataFrame({"region": ["Lagos", "Abuja"], "manager": ["Tunde", "Amaka"]})

# Perform the merge
merged_df = df.merge(managers, on="region", how="left")   # inner, left, right, outer

# Define df1 and df2 to demonstrate concat
df1 = pd.DataFrame({"region": ["Kano"], "product": ["B"], "rev": [150], "cost": [100], "profit": [50], "margin": [0.33]})
df2 = pd.DataFrame({"region": ["Enugu"], "product": ["A"], "rev": [250], "cost": [180], "profit": [70], "margin": [0.28]})

# Stack tables vertically
concatenated_df = pd.concat([df1, df2])
concatenated_df
```
